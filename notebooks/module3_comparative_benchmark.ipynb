# Module 3: Comparative Benchmark (LSTM vs GRU vs BiLSTM vs GCN+LSTM)
### Research Project: *Prediction Of Drug Dosage Of Diabetic Patients Based On Ontology Mapping And Deep Learning Models*

This notebook trains and evaluates all 4 architectures under identical data splits, random seeds, and loss weighting, outputting a comparative research report.

**Architectures Compared:**
1. `LSTM` (Temporal Baseline)
2. `GRU` (Temporal Baseline)
3. `BiLSTM` (Temporal Baseline)
4. `GCN_LSTM` (Proposed Ontology-Grounded Graph Architecture)

```python
!pip install -q torch-geometric pandas numpy scikit-learn matplotlib pyyaml

import os
import sys
import torch
import pandas as pd
import matplotlib.pyplot as plt

repo_root = os.path.abspath(".")
if repo_root not in sys.path:
    sys.path.append(repo_root)

device = "cuda" if torch.cuda.is_available() else "cpu"
print(f"Executing comparative benchmark on: {device}")
```

```python
from module3.config import ExperimentConfig, TrainConfig, ModelConfig
from module3.pipeline.experiment_runner import ExperimentRunner

# Setup benchmark configuration
train_cfg = TrainConfig(
    batch_size=16,
    epochs=10,
    learning_rate=1e-3,
    loss_fn="weighted_smooth_l1",
    pos_weight=8.0,
    early_stopping_patience=5,
    device=device,
)

exp_cfg = ExperimentConfig(train=train_cfg)
runner = ExperimentRunner(exp_cfg)

# Run benchmark across all 4 architectures
results = runner.run_comparative_benchmark(
    model_types=["lstm", "gru", "bilstm", "gcn_lstm"],
    epochs=10,
    batch_size=16,
)
```
